```python
# A* for 8-puzzle - CASE 2: g(n) = depth, h(n) = Manhattan distance
# State = tuple of 9 numbers (row by row), 0 = blank
START = (2, 8, 3,
         1, 6 ,4 ,
         0, 7, 5)
GOAL  = (1, 2, 3,
         8, 0, 4,
         7, 6, 5)


def h(state):
    """Sum of Manhattan distances of all tiles (blank not counted)."""
    total = 0
    for i in range(9):
        tile = state[i]
        if tile != 0:
            cur_row, cur_col = i // 3, i % 3
            goal_pos = GOAL.index(tile)
            goal_row, goal_col = goal_pos // 3, goal_pos % 3
            total += abs(cur_row - goal_row) + abs(cur_col - goal_col)
    return total


def get_children(state):
    """Move the blank Up, Down, Left, Right."""
    children = []
    b = state.index(0)
    row, col = b // 3, b % 3
    moves = [("Up", -1, 0), ("Down", 1, 0), ("Left", 0, -1), ("Right", 0, 1)]
    for name, dr, dc in moves:
        r, c = row + dr, col + dc
        if 0 <= r < 3 and 0 <= c < 3:
            new = list(state)
            n = r * 3 + c
            new[b], new[n] = new[n], new[b]
            children.append((name, tuple(new)))
    return children


def print_state(state):
    for i in range(0, 9, 3):
        print(" ".join("_" if x == 0 else str(x) for x in state[i:i + 3]))
    print()


def is_solvable(state):
    t = [x for x in state if x != 0]
    g = [x for x in GOAL if x != 0]
    inv = lambda a: sum(a[i] > a[j] for i in range(8) for j in range(i + 1, 8))
    return inv(t) % 2 == inv(g) % 2


def astar(start):
    # OPEN: state -> (f, g, h); CLOSED: set of expanded states
    open_list = {start: (h(start), 0, h(start))}
    closed = set()
    parent = {start: (None, None)}
    expanded = 0

    while open_list:
        # pick node with lowest f (tie -> lower h)
        n = min(open_list, key=lambda s: (open_list[s][0], open_list[s][2]))
        f, g, hn = open_list.pop(n)
        closed.add(n)
        expanded += 1

        if n == GOAL:                       # goal test
            path, moves = [], []
            while n is not None:
                path.append((n, parent[n][1]))
                n = parent[n][0]
            return path[::-1], expanded

        for move, child in get_children(n):
            if child in closed:
                continue
            gc = g + 1
            hc = h(child)
            fc = gc + hc
            if child not in open_list or gc < open_list[child][1]:
                open_list[child] = (fc, gc, hc)
                parent[child] = (n, move)
    return None, expanded


if __name__ == "__main__":
    print("CASE 2: h(n) = Manhattan distance")
    print("Start state:"); print_state(START)
    print("Goal state:");  print_state(GOAL)
    print("h(start) =", h(START), "\n")

    if not is_solvable(START):
        print("This puzzle is NOT solvable.")
    else:
        path, expanded = astar(START)
        for step, (state, move) in enumerate(path):
            g = step
            print(f"Step {step}  move={move}  g={g}  h={h(state)}  f={g + h(state)}")
            print_state(state)
        print("Total moves   :", len(path) - 1)
        print("Nodes expanded:", expanded)
```
