```python

# FOL Resolution - John likes peanuts

# Knowledge Base (Facts)
food = ["apple", "vegetables"]
eats = [("Anil", "peanuts")]
alive = ["Anil"]

# Rule 1: Alive means not killed
not_killed = []

for person in alive:
    not_killed.append(person)

# Rule 2: Anything eaten by a person
# who is not killed is food
for person, item in eats:
    if person in not_killed:
        food.append(item)

# Rule 3: John likes all food
likes = []

for item in food:
    likes.append(item)

# Prove the query
query = "peanuts"

if query in likes:
    print("Query proved: John likes peanuts")
else:
    print("Query not proved")

print("Food:", food)
print("John likes:", likes)
```
