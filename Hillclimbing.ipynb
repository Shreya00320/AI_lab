```python
"""
Hill Climbing Search for the N-Queens problem (default N = 4)

State      : list x, where x[i] = row of the queen in column i (one queen per column)
Neighbour  : swap the row positions of two queens (x[i] <-> x[j])
Cost       : number of pairs of queens attacking each other (same row or same diagonal)
Tie-break  : if several neighbours have the same cost, pick the one whose swapped
             pair has the smallest column numbers, i.e. (0,1) before (0,2) before (1,2) ...
Stop       : when the best neighbour is NOT better than the current state
"""
import random


def cost(state):
    """Number of attacking pairs of queens."""
    n = len(state)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            same_row = state[i] == state[j]
            same_diag = abs(state[i] - state[j]) == j - i
            if same_row or same_diag:
                attacks += 1
    return attacks


def attacking_pairs(state):
    """List the attacking pairs (for showing the cost calculation)."""
    n = len(state)
    pairs = []
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                pairs.append((i, j))
    return pairs


def print_board(state):
    n = len(state)
    for row in range(n):
        print("   " + " ".join("Q" if state[col] == row else "." for col in range(n)))


def hill_climbing(initial, verbose=True):
    current = list(initial)
    current_cost = cost(current)
    step = 0

    if verbose:
        print(f"Initial state : x = {current}")
        print(f"Initial cost  : {current_cost}  (attacking pairs: {attacking_pairs(current)})")
        print_board(current)

    while True:
        if current_cost == 0:
            if verbose:
                print("\nGoal reached: no pair of queens is attacking.")
            return current, current_cost

        step += 1
        if verbose:
            print(f"\n================ Step {step} ================")
            print(f"Current state x = {current}, cost = {current_cost}")
            print("Neighbours (swap columns i and j):")

        best_neighbour, best_cost, best_pair = None, None, None
        n = len(current)
        for i in range(n):
            for j in range(i + 1, n):
                neighbour = list(current)
                neighbour[i], neighbour[j] = neighbour[j], neighbour[i]
                c = cost(neighbour)
                if verbose:
                    print(f"   swap (x{i}, x{j}) -> {neighbour}  cost = {c}"
                          f"  attacking pairs = {attacking_pairs(neighbour)}")
                # strict '<' keeps the FIRST (smallest column pair) on ties
                if best_cost is None or c < best_cost:
                    best_neighbour, best_cost, best_pair = neighbour, c, (i, j)

        if verbose:
            print(f"Best neighbour: swap (x{best_pair[0]}, x{best_pair[1]}) "
                  f"-> {best_neighbour} with cost {best_cost}")

        if best_cost >= current_cost:
            if verbose:
                print("Best neighbour is not better than current state -> STOP "
                      "(local minimum).")
            return current, current_cost

        current, current_cost = best_neighbour, best_cost
        if verbose:
            print(f"Move to {current}")
            print_board(current)


if __name__ == "__main__":
    # ---- Given problem: x0 = 3, x1 = 1, x2 = 2, x3 = 0 ----
    final_state, final_cost = hill_climbing([3, 1, 2, 0])
    print("\n----------------- RESULT -----------------")
    print(f"Final state : {final_state}")
    print(f"Final cost  : {final_cost}")
    print_board(final_state)

    # ---- Optional: random initial state with random restarts ----
    # Hill climbing can get stuck at a local minimum, so restart until a goal is found.
    print("\n=========== Random restart version ===========")
    N = 4
    restarts = 0
    while True:
        restarts += 1
        start = random.sample(range(N), N)      # random permutation of rows
        state, c = hill_climbing(start, verbose=False)
        if c == 0:
            break
    print(f"Solution found after {restarts} restart(s): {state}")
    print_board(state)
```
